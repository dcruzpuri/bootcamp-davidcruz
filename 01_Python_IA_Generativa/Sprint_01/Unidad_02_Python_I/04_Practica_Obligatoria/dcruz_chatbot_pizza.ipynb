```python
# Saludo inicial y pedido de pizzas 
print("Buenas tardes, bienvenido al servicio de pedido online, el precio de nuestras pizzas es de 8,95 para la familiar o de 9,90 para la mediana.")


# Nos aseguramos de que el usuario ingrese un número válido de pizzas
pizza_familiar = int(input("Ingrese la cantidad de pizzas familiares: "))
pizza_mediana = int(input("Ingrese la cantidad de pizzas medianas: "))

# Confirmación del pedido y petición de dirección
direcc = input(f"Estupendo, se están preparando {pizza_familiar} pizzas familiares y {pizza_mediana} pizzas medianas. Dígame su dirección: ")

# Cálculo del monto total
total_pizzas = pizza_familiar + pizza_mediana
monto_total = (pizza_familiar * 8.95) + (pizza_mediana * 9.90)

# Respuesta final
print(f"Le mandaremos las {total_pizzas} pizzas a la dirección {direcc}. Serán {monto_total:.2f}€. Muchas gracias por su pedido.")
```
